```python
import cv2
```

```python
video_path = "test/video.mp4"
```

```python
cap = cv2.VideoCapture(video_path)
```

```python
if not cap.isOpened():
    print("Error opening video")
else:
    print("Video opened successfully")
```

```python
fps = cap.get(cv2.CAP_PROP_FPS)
fps
```

```python
total_frames = cap.get(cv2.CAP_PROP_FRAME_COUNT)
total_frames
```

```python
total_frames / fps
```

```python
cap.get(cv2.CAP_PROP_FRAME_WIDTH)
```

```python
cap.get(cv2.CAP_PROP_FRAME_HEIGHT)
```

```python
cap = cv2.VideoCapture(video_path)
frame_count = 0

while True:
    ret, frame = cap.read()
    if not ret:
        break

    cv2.imwrite(f"test/frame/frame_{frame_count}.jpg", frame)
    frame_count += 1
    
cap.release()
```

```python
cap = cv2.VideoCapture(video_path)

while True:
    ret, frame = cap.read()
    if not ret:
        break
    
    cv2.imshow("video", frame)

    if cv2.waitKey(25) == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
