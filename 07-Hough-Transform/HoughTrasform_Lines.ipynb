```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("D:\Computer-Vision\Image-Processing\HoughTransform\sudoku.jpg")
```

```python
plt.figure(figsize=(12,6))
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.show()
```

Extracting Edges.

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
blure = cv2.GaussianBlur(gray,(5,5),0)
edges = cv2.Canny(gray, 200, 250)
```

```python
plt.figure(figsize=(12,6))
plt.imshow(edges, cmap='gray')
plt.show()
```

Hough line. A method in cv2 for extracting line.

edges >>> extracted by Canny. 1 >>> step of distance. as less as ditance be we will have more accuracy in detecting lines. np.pi/180 >>> move on degree 1 degree in every step. 200 >>> treshold >>> encounter times for accept the line structure. if you wants to detect small lines you have to set it lower.

```python
lines = cv2.HoughLines(edges, 1, np.pi/180, 150)
```

```python
lines[0]
```

```python
image_line1 = image.copy()

for line in lines:
    rho, theta = line[0]
    a = np.cos(theta)
    b = np.sin(theta)
    x0 = a * rho
    y0 = b * rho
    x1 = int(x0 + 1000 * (-b))
    y1 = int(y0 + 1000 * (a))
    x2 = int(x0 - 1000 * (-b))
    y2 = int(y0 - 1000 * (a))
    cv2.line(image_line1, (x1, y1), (x2, y2), (0, 0, 255), 2)
```

```python
plt.imshow(cv2.cvtColor(image_line1, cv2.COLOR_BGR2RGB))
plt.show()
```

Probabilistic Hough Transform >>> Givving the line cordinate directly. 

Parameters: edges >>> input. 1 >>> step of distance. as less as ditance be we will have more accuracy in detecting lines. np.pi/180 >>> move on degree 1 degree in every step.
60 >>> treshold >>> encounter times for accept the line structure. minLineLength=100 >>> smallest line. maxLineGap=60 >>> The maximum distance of a line that is split in two.

```python
lines = cv2.HoughLinesP(edges, 1, np.pi / 180, threshold=60, minLineLength=100, maxLineGap=60)
```

```python
lines[0]
```

```python
image_line2 = image.copy()

for line in lines:
    x1, y1, x2, y2 = line[0]
    cv2.line(image_line2, (x1, y1), (x2, y2), (0, 0, 255), 2)
```

```python
plt.figure(figsize=(12,6))
plt.imshow(cv2.cvtColor(image_line2, cv2.COLOR_BGR2RGB))
plt.show()
```
